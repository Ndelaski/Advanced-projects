<p style="text-align: justify;">
    The main scope of the project is to improve the accuracy of diagnosing Long COVID-19 using Bayesian 
Networks. The project aims to reproduce and enhance the results of a previous study (DOI: https://doi.org/10.1101/2020.10.22.20217554) that used 
Bayesian Networks for diagnosing COVID-19. Additionally, the project aims to test the 
performance of the Bayesian Network model on patient data from a hospital in Sydney.
</p>

In a recent study, Butcher and Fenton \cite{butcher2020extending} proposed a Bayesian network model to predict the diagnosis of COVID-19 based on a range of symptoms. The model was able to accurately classify patients as positive or negative for COVID-19 based on their symptoms.

To implement the Bayesian network, we used the `pgmpy` library \cite{ankan2015pgmpy}, which provides tools for building and analyzing probabilistic graphical models in Python.
