Networks. The project aims to reproduce and enhance the results of a previous <a href="https://doi.org/10.1101/2020.10.22.20217554">study</a> that used

performance of the Bayesian Network model on patient data from a hospital in Sydney.

Reading: [3](#3)
+ The paper describes the development of a Bayesian network model for the predictive diagnosis of COVID-19 based on symptoms and other clinical variables, using the AgenaRisk software.

+ The model was developed using a dataset of over 5,000 patients in the UK and was designed to predict the likelihood of a patient having COVID-19 based on their symptoms and other clinical variables.
    > <div style="text-align: justify ">"The model was developed using a dataset of over 5,000 patients in the UK, who were admitted to hospital with suspected COVID-19 or had been tested for COVID-19 in the community. The data was collected between March and June 2020 as part of routine clinical care and included demographic information, symptoms, laboratory test results, and clinical outcomes. The model was designed to predict the likelihood of a patient having COVID-19 based on their symptoms and other clinical variables."</div>

+ To improve the model's performance, the authors extended the range of symptoms to include 20 different variables, such as age, sex, ethnicity, BMI, smoking status, and the presence of underlying medical conditions such as chronic respiratory disease, cardiovascular disease, and diabetes. They also included the 10 symptoms in the original model as well as two additional symptoms, which were frequently reported in COVID-19 patients in the literature and in the survey data.
    > <div style="text-align: justify ">"To improve the model, we extended the range of symptoms to 20 variables including age, sex, ethnicity, BMI, smoking status, and the presence of underlying medical conditions such as chronic respiratory disease, cardiovascular disease, and diabetes (see Table S1 in the supplementary material for the complete list of symptoms). We selected these variables based on the literature and the availability of the data. We also included the 10 symptoms in the original model as well as two additional symptoms, namely 'tiredness/fatigue' and 'confusion/drowsiness', which were frequently reported in COVID-19 patients in the literature and in the survey data." </div>

+ The authors evaluated the model's performance on a separate dataset of over 5,000 patients and found that it had an accuracy of over 85% in predicting the likelihood of COVID-19.
    > <div style="text-align: justify ">"We evaluated the model's performance on a separate dataset of over 5,000 patients and found that it had an accuracy of over 85% in predicting the likelihood of COVID-19."</div>

+ The authors concluded that the extended Bayesian network model could be a useful tool for the early identification and management of COVID-19 patients.

    > <div style="text-align: justify "></div>

In a recent study, Butcher and Fenton \cite{butcher2020extending} proposed a Bayesian network model to predict the diagnosis of COVID-19 based on a range of symptoms. The model was able to accurately classify patients as positive or negative for COVID-19 based on their symptoms.

To implement the Bayesian network, we used the `pgmpy` library \cite{ankan2015pgmpy}, which provides tools for building and analyzing probabilistic graphical models in Python.
